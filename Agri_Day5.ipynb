```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("patelris/crop-yield-prediction-dataset")

print("Path to dataset files:", path)
```

```python
import torch

dev = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {dev}")
```

```python
x_data = [1.0, 2.0, 3.0, 4.0]
y_data = [2.0, 4.0, 6.0, 8.0]

x = torch.tensor(x_data, dtype=torch.float32, device=dev)
y = torch.tensor(y_data, dtype=torch.float32, device=dev)
```

```python
w = torch.zeros(1, device=dev, requires_grad=True)
b = torch.zeros(1, device=dev, requires_grad=True)
```

```python
y_pred = w * x + b
loss = ((y_pred - y) ** 2).mean()
loss.backward()

print("Loss:", loss.item())
print("w.grad:", w.grad)
print("b.grad:", b.grad)
```
