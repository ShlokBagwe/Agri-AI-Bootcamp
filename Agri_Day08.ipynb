```python
import glob
import kagglehub
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader

device = torch.device("cpu")
print("Using device:", device)

# Download dataset via kagglehub (no manual file upload needed)
path = kagglehub.dataset_download("atharvaingle/crop-recommendation-dataset")
csv_files = glob.glob(f"{path}/**/*.csv", recursive=True)
df = pd.read_csv(csv_files[0])

print("Dataset shape:", df.shape)
print(df.head(3))
```

```python
# Separate numeric features and encode crop label as integer class IDs
feature_cols = [c for c in df.columns if c != "label"]
X_data = torch.tensor(df[feature_cols].values, dtype=torch.float32)

classes = sorted(df["label"].unique())
label_to_idx = {name: idx for idx, name in enumerate(classes)}
y_data = torch.tensor(df["label"].map(label_to_idx).values, dtype=torch.long)


class FarmDS(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


dataset = FarmDS(X_data, y_data)
loader = DataLoader(dataset, batch_size=32, shuffle=True)

print(f"Total rows: {len(dataset)} | Total batches per epoch: {len(loader)}")
```

```python
for batch_idx, (xb, yb) in enumerate(loader):
    xb, yb = xb.to(device), yb.to(device)
    print(f"Batch {batch_idx:2d} | X shape: {tuple(xb.shape)} | y shape: {tuple(yb.shape)}")
```
